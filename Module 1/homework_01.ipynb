```python
x = df[df['origin'] == 'Asia'][['vehicle_weight', 'model_year']].head(7).values
x
```

```python
xt = x.T
```

```python
xt
```

```python
xtx = np.matmul(xt, x)
xtx
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
y
```

```python
# inverse of XTX
xtx_inv = np.linalg.inv(xtx)
```

```python
w = np.dot(np.dot(xtx_inv, x.T), y)
```

```python
print(w)
```

```python
print(w[0]+w[1])
```
