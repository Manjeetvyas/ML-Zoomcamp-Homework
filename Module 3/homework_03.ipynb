## Validation framework

```python
#setting up validation framework (TRAIN, VALIDATION, SPLIT) using scikit-learn in (60:20:20):

from sklearn.model_selection import train_test_split
```

```python
len(df_train), len(df_test), len(df_val)
```

```python
df_train = df_train.reset_index(drop=True)
df_val = df_val.reset_index(drop=True)
df_test = df_test.reset_index(drop=True)
```

```python
y_train = df_train.converted.values
y_val = df_val.converted.values
y_test = df_test.converted.values
```

```python
del df_train["converted"]
del df_val["converted"]
del df_test["converted"]
```

# Question 6

```python
C =  [0.000001, 0.00001, 0.0001, 0.001]
```

```python
def calc_pred(C):
    from sklearn.feature_extraction import DictVectorizer
    train_dicts = df_train[categorical + numerical].to_dict(orient='records')
    dv = DictVectorizer(sparse=False)
    X_train = dv.fit_transform(train_dicts)
    val_dicts = df_val[categorical + numerical].to_dict(orient='records')
    X_val = dv.transform(val_dicts)
    from sklearn.linear_model import LogisticRegression
    model = LogisticRegression(solver='liblinear', C=C, max_iter=1000, random_state=42)
    model.fit(X_train, y_train)
    y_pred = model.predict_proba(X_val)[:,1]
    convert_decision = (y_pred >= 0.5)
    pred = (y_val == convert_decision).mean().round(3)
    return pred
```

```python
for i in C:
    print(calc_pred(i))
```

```python
# Q-> Which of these C leads to the best accuracy on the validation set?

# 0.000001
# 0.00001
# 0.0001
# 0.001
# Note: If there are multiple options, select the smallest C.
```

```python
# Ans-> C=  0.001, returns 0.645
```
